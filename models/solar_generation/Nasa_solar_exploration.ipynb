```python
# NASA POWER API endpoint
BASE_URL = "https://power.larc.nasa.gov/api/temporal/daily/point"

# Test location: Nairobi, Kenya
test_location = {
    "name": "Nairobi",
    "lat": -1.2921,
    "lon": 36.8219
}

# Parameters we need
parameters = [
    "ALLSKY_SFC_SW_DWN",  # Solar irradiance
    "T2M",                 # Temperature
    "WS10M",               # Wind Speed
]

# Last 30 days
end_date = datetime.now()
start_date = end_date - timedelta(days=30)

params = {
    "parameters": ",".join(parameters),
    "community": "RE",
    "longitude": test_location["lon"],
    "latitude": test_location["lat"],
    "start": start_date.strftime('%Y%m%d'),
    "end": end_date.strftime('%Y%m%d'),
    "format": "JSON"
}

print(f"Testing API for {test_location['name']}...")
print(f"Date range: {params['start']} to {params['end']}")

response = requests.get(BASE_URL, params=params)
print(f"\nStatus Code: {response.status_code}")

if response.status_code == 200:
    data = response.json()
    print("✓ API call successful!")
    print(f"\nParameters returned: {list(data['properties']['parameter'].keys())}")
else:
    print(f"✗ Error: {response.text}")
```
